dimensionality reduction using PCA

standardize the dataset

```python
from sklearn.preprocessing import StandardScaler

# Standardize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

print("Shape of Scaled Data:", scaled_data.shape)
```

apply PCA

```python
from sklearn.decomposition import PCA

# Apply PCA
pca = PCA()

pca_data = pca.fit_transform(scaled_data)
```

expline variance ratio

```python
explained_variance = pca.explained_variance_ratio_

print(explained_variance)
```

screen plot

```python
plt.figure(figsize=(10,6))

plt.plot(range(1,len(explained_variance)+1),explained_variance, marker='o')

plt.xlabel("Principal Components")
plt.ylabel("Explained Variance")
plt.title("Scree Plot")

plt.grid(True)
plt.show()
```

cumulative explianed variance

```python
cumulative_variance = explained_variance.cumsum()

plt.figure(figsize=(10,6))

plt.plot(
    range(1,len(cumulative_variance)+1),
    cumulative_variance,
    marker='o'
)

plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")

plt.axhline(0.95,color='red',linestyle='--')

plt.grid(True)
plt.show()
```

optimal components

```python
optimal_components = np.argmax(cumulative_variance >=0.82)+1
print("Optimal Number of Components:", optimal_components)
```

transform dataset

```python
pca = PCA(n_components=optimal_components)

pca_dataset = pca.fit_transform(scaled_data)

print(pca_dataset.shape)
```

clustering with original data

k -means

```python
from sklearn.cluster import KMeans

kmeans_original = KMeans(n_clusters=5, random_state=40, n_init=10)

original_labels = kmeans_original.fit_predict(scaled_data)
```

visualize clusters

```python
plt.figure(figsize=(8,6))

plt.scatter(
    scaled_data[:,0],
    scaled_data[:,1],
    c=original_labels,
    cmap='viridis'
)

plt.xlabel(df.columns[0])
plt.ylabel(df.columns[1])

plt.title("Clusters on Original Dataset")

plt.show()
```

evaluation of clustering

```python
from sklearn.metrics import silhouette_score
from sklearn.metrics import davies_bouldin_score

silhouette_original = silhouette_score(
    scaled_data,
    original_labels
)

db_original = davies_bouldin_score(
    scaled_data,
    original_labels
)

print("Silhouette Score :", silhouette_original)

print("Davies-Bouldin Index :", db_original)
```

clustering with PCA data

 aplly k-means

```python
kmeans_pca = KMeans(
    n_clusters=4,
    random_state=40,
    n_init=10
)

pca_labels = kmeans_pca.fit_predict(pca_dataset)
```

visualize PCA clusters

```python
plt.figure(figsize=(10,7))

plt.scatter(
    pca_dataset[:,0],
    pca_dataset[:,1],
    c=pca_labels,
    cmap='rainbow'
)

plt.xlabel("Principal Component A")
plt.ylabel("Principal Component B")

plt.title("Clusters after PCA")

plt.show()
```

evaluate PCA clustering

```python
silhouette_pca = silhouette_score( pca_dataset,pca_labels)

db_pca = davies_bouldin_score(pca_dataset, pca_labels)

print("Silhouette Score :", silhouette_pca)

print("Davies-Bouldin Index :", db_pca)
```

comparison ansd analysis

creating comparison table

```python
comparison = pd.DataFrame({

    "Method":[
        "Original Dataset",
        "PCA Dataset"
    ],

    "Silhouette Score":[
        silhouette_original,
        silhouette_pca
    ],

    "Davies-Bouldin Index":[
        db_original,
        db_pca
    ]

})

comparison
```

camparing using bar charts

```python
comparison.plot(
    x="Method",
    y="Silhouette Score",
    kind="bar",
    figsize=(10,6)
)

plt.title("Silhouette Score Comparison")

plt.show()
```

davies-bouldin index

```python
comparison.plot(
    x="Method",
    y="Davies-Bouldin Index",
    kind="bar",
    figsize=(10,7)
)

plt.title("Davies-Bouldin Index Comparison")

plt.show()
```
